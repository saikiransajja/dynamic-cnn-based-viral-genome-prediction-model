```python
import numpy as np
from sklearn.model_selection import train_test_split, KFold
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, average_precision_score,
                             roc_auc_score, confusion_matrix, matthews_corrcoef)
from Bio import SeqIO
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (Conv1D, MaxPooling1D, Dense, Dropout,
                                     GlobalAveragePooling1D)
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.utils import to_categorical
```

```python
# Function to get the index of a 4-mer, skipping if it contains 'N'
def get_4mer_index(kmer):
    bases = ['A', 'C', 'G', 'T']
    index = 0
    for i, base in enumerate(kmer):
        if base not in bases:
            return None
        index += bases.index(base) * (4 ** (3 - i))  # 4^3 = 64, 4^2 = 16, etc.
    return index

# Function to convert a 4-mer into a one-hot encoded vector of size 256, skipping 'N'
def one_hot_4mer(kmer):
    vector = np.zeros(256)
    index = get_4mer_index(kmer)
    if index is not None:
        vector[index] = 1
    return vector

# Function to generate the one-hot encoded matrix for a single sequence
def one_hot_encoding_matrix(sequence):
    kmers = [sequence[i:i+4] for i in range(len(sequence) - 3)]
    matrix = np.array([one_hot_4mer(kmer) for kmer in kmers if get_4mer_index(kmer) is not None])
    return matrix

# Load your combined sequences and labels from FASTA files
def load_fasta_files(fasta_files, predefined_labels):
    sequences = []
    labels = []
    for fasta_file, label in zip(fasta_files, predefined_labels):
        for record in SeqIO.parse(fasta_file, "fasta"):
            sequences.append(str(record.seq))
            labels.append(label)
    return sequences, labels

# Preprocess the sequences
def preprocess_sequence(sequence):
    return one_hot_encoding_matrix(sequence)

# Build the CNN model
def build_dynamic_cnn(input_shape, num_classes):
    model = Sequential()
    model.add(Conv1D(filters=32, kernel_size=3, activation='relu', input_shape=input_shape))
    model.add(MaxPooling1D(pool_size=2))
    model.add(Conv1D(filters=64, kernel_size=3, activation='relu'))
    model.add(MaxPooling1D(pool_size=2))
    model.add(Conv1D(filters=128, kernel_size=3, activation='relu'))
    model.add(MaxPooling1D(pool_size=2))
    model.add(GlobalAveragePooling1D())  # For variable-length sequences
    model.add(Dense(128, activation='relu'))
    model.add(Dropout(0.5))
    model.add(Dense(num_classes, activation='softmax'))
    return model

# Train the model one sequence at a time
def train_model_on_sequence(model, sequence, label):
    sequence = np.expand_dims(sequence, axis=0)  # Reshape to (1, sequence_length, 256)
    label = np.expand_dims(label, axis=0)  # Reshape to (1, num_classes)
    model.train_on_batch(sequence, label)

# Evaluate the model one sequence at a time
def evaluate_model_on_sequence(model, sequence, label):
    sequence = np.expand_dims(sequence, axis=0)
    y_pred_probs = model.predict(sequence)
    y_pred = np.argmax(y_pred_probs, axis=1)
    return y_pred_probs, y_pred

# Calculate specificity
def calculate_specificity(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    true_negatives = cm.sum(axis=1) - np.diag(cm)
    false_positives = cm.sum(axis=0) - np.diag(cm)
    
    denom = true_negatives + false_positives
    specificity = np.where(denom != 0, true_negatives / denom, 0.0)
    
    return np.mean(specificity)

def calculate_mcc_from_confusion_matrix(confusion_matrix):
    # Flatten the confusion matrix to calculate MCC
    flattened_y_true = []
    flattened_y_pred = []

    # Extract true and predicted values from the confusion matrix
    for i in range(len(confusion_matrix)):
        for j in range(len(confusion_matrix)):
            flattened_y_true.extend([i] * confusion_matrix[i][j])
            flattened_y_pred.extend([j] * confusion_matrix[i][j])

    # Calculate MCC
    mcc = matthews_corrcoef(flattened_y_true, flattened_y_pred)
    return mcc

# Combine all your FASTA file names and corresponding labels
fasta_files = ['covid.fasta', 'dengue.fasta', 'hepatitis.fasta', 'influenza.fasta', 'mers.fasta']
predefined_labels = ['COVID', 'Dengue', 'Hepatitis', 'Influenza', 'MERS']

# Load data
sequences, labels = load_fasta_files(fasta_files, predefined_labels)

# Shuffle the data
data = list(zip(sequences, labels))
np.random.shuffle(data)
sequences, labels = zip(*data)

# Encode labels
label_encoder = LabelEncoder()
encoded_labels = label_encoder.fit_transform(labels)
num_classes = len(np.unique(encoded_labels))
y = to_categorical(encoded_labels)

# Split data into train and test sets
train_sequences, test_sequences, train_labels, test_labels = train_test_split(sequences, y, test_size=0.2, random_state=42)

# Build the model
input_shape = (None, 256)  # Variable-length input for sequences
model = build_dynamic_cnn(input_shape, num_classes)
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
```

```python
# Cross-validation Phase
print("\nCross-validation phase:")
kf = KFold(n_splits=4, shuffle=True, random_state=42)
all_metrics = []
confusion_matrices = []

# Create lists to store fold metrics
fold_metrics_summary = []

for fold, (train_index, val_index) in enumerate(kf.split(train_sequences)):
    print(f"\nFold {fold + 1}/4")
    
    fold_train_sequences = np.array(train_sequences)[train_index]
    fold_train_labels = np.array(train_labels)[train_index]
    fold_val_sequences = np.array(train_sequences)[val_index]
    fold_val_labels = np.array(train_labels)[val_index]
    
    # Train on the fold
    for sequence_number, seq in enumerate(fold_train_sequences):
        sequence = preprocess_sequence(seq)
        
        # Ensure the sequence is not empty or malformed
        if sequence is None or len(sequence) == 0 or len(sequence.shape) < 2:
            print(f"Skipping training sequence {sequence_number + 1}/{len(fold_train_sequences)} due to invalid shape")
            continue
        if sequence.shape[1] == 0:
            print(f"Skipping training sequence {sequence_number + 1}/{len(fold_train_sequences)} due to zero length")
            continue
        
        label = fold_train_labels[sequence_number]
        train_model_on_sequence(model, sequence, label)

        # Print progress of training
        print(f"Trained on sequence {sequence_number + 1}/{len(fold_train_sequences)}")

    # Evaluate on the validation set
    all_y_true = []
    all_y_pred = []
    all_y_pred_probs = []
    total_val_sequences = len(fold_val_sequences)
    
    for sequence_number, seq in enumerate(fold_val_sequences):
        sequence = preprocess_sequence(seq)
        
        # Ensure the sequence is not empty or malformed
        if sequence is None or len(sequence) == 0 or len(sequence.shape) < 2:
            print(f"Skipping validation sequence {sequence_number + 1}/{total_val_sequences} due to invalid shape")
            continue
        if sequence.shape[1] == 0:
            print(f"Skipping validation sequence {sequence_number + 1}/{total_val_sequences} due to zero length")
            continue
        
        label = np.argmax(fold_val_labels[sequence_number])
        y_pred_probs, y_pred = evaluate_model_on_sequence(model, sequence, label)

        if y_pred is not None and y_pred_probs is not None:
            all_y_true.append(label)
            all_y_pred.append(y_pred[0])
            all_y_pred_probs.append(y_pred_probs[0])

        # Print progress of validation
        print(f"Evaluated validation sequence {sequence_number + 1}/{total_val_sequences}")

    # Calculate metrics for the fold
    accuracy = accuracy_score(all_y_true, all_y_pred)
    precision = precision_score(all_y_true, all_y_pred, average='weighted')
    recall = recall_score(all_y_true, all_y_pred, average='weighted')  # Sensitivity
    f1 = f1_score(all_y_true, all_y_pred, average='weighted')
    specificity = calculate_specificity(all_y_true, all_y_pred)

    # For AUROC and AUPRC, convert the predictions and true values to one-hot encoded format
    all_y_true_onehot = to_categorical(all_y_true, num_classes=num_classes)
    auroc = roc_auc_score(np.array(all_y_true_onehot), np.array(all_y_pred_probs), multi_class='ovr')
    auprc = average_precision_score(np.array(all_y_true_onehot), np.array(all_y_pred_probs), average='macro')

    # Store fold metrics in summary
    fold_metrics_summary.append((accuracy, precision, recall, f1, specificity, auroc, auprc))

    # Print fold metrics
    print(f"Fold {fold + 1} Metrics:")
    print(f"Accuracy: {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall (Sensitivity): {recall:.4f}")
    print(f"F1 Score: {f1:.4f}")
    print(f"Specificity: {specificity:.4f}")
    print(f"AUROC: {auroc:.4f}")
    print(f"AUPRC: {auprc:.4f}")

    all_metrics.append((accuracy, precision, recall, f1, specificity, auroc, auprc))
    confusion_matrices.append(confusion_matrix(all_y_true, all_y_pred))

# Calculate test metrics
all_y_true_test = []
all_y_pred_test = []
all_y_pred_probs_test = []
total_test_sequences = len(test_sequences)

for sequence_number, seq in enumerate(test_sequences):
    sequence = preprocess_sequence(seq)

    # Ensure the sequence is not empty or malformed
    if sequence is None or len(sequence) == 0 or len(sequence.shape) < 2:
        print(f"Skipping test sequence {sequence_number + 1}/{total_test_sequences} due to invalid shape")
        continue
    if sequence.shape[1] == 0:
        print(f"Skipping test sequence {sequence_number + 1}/{total_test_sequences} due to zero length")
        continue
    
    label = np.argmax(test_labels[sequence_number])
    y_pred_probs, y_pred = evaluate_model_on_sequence(model, sequence, label)

    if y_pred is not None and y_pred_probs is not None:
        all_y_true_test.append(label)
        all_y_pred_test.append(y_pred[0])
        all_y_pred_probs_test.append(y_pred_probs[0])

    # Print progress of testing
    print(f"Evaluated test sequence {sequence_number + 1}/{total_test_sequences}")

# Calculate test metrics
test_accuracy = accuracy_score(all_y_true_test, all_y_pred_test)
test_precision = precision_score(all_y_true_test, all_y_pred_test, average='weighted')
test_recall = recall_score(all_y_true_test, all_y_pred_test, average='weighted')  # Sensitivity
test_f1 = f1_score(all_y_true_test, all_y_pred_test, average='weighted')
test_specificity = calculate_specificity(all_y_true_test, all_y_pred_test)
all_y_true_onehot_test = to_categorical(all_y_true_test, num_classes=num_classes)
test_auroc = roc_auc_score(np.array(all_y_true_onehot_test), np.array(all_y_pred_probs_test), multi_class='ovr')
test_auprc = average_precision_score(np.array(all_y_true_onehot_test), np.array(all_y_pred_probs_test), average='macro')
```

```python
# List to store MCCs for each fold
mcc_values = []

# Calculate MCC for each confusion matrix and store it
for i in range(len(confusion_matrices)):
    cm = confusion_matrices[i]  # Get the confusion matrix directly
    mcc = calculate_mcc_from_confusion_matrix(cm)  # Pass the confusion matrix directly
    mcc_values.append(mcc)  # Store MCC for this fold
    print(f"Matthews Correlation Coefficient (MCC) for Fold {i + 1}: {mcc}")
    

final_confusion_matrix = confusion_matrix(all_y_true_test, all_y_pred_test)
test_mcc = calculate_mcc_from_confusion_matrix(final_confusion_matrix)


```

```python
# Print test metrics
print(f"\nTest Set Metrics:")
print(f"Test Accuracy: {test_accuracy:.4f}")
print(f"Test Precision: {test_precision:.4f}")
print(f"Test Recall: {test_recall:.4f}")
print(f"Test F1 Score: {test_f1:.4f}")
print(f"Test Specificity: {test_specificity:.4f}")
print(f"Test AUROC: {test_auroc:.4f}")
print(f"Test AUPRC: {test_auprc:.4f}")
print(f"Test MCC: {test_mcc:.4f}")

# Include test metrics in the average calculations
all_metrics.append((test_accuracy, test_precision, test_recall, test_f1, test_specificity, test_auroc))

# Calculate average metrics across all folds and include test metrics
avg_metrics = np.mean(all_metrics, axis=0)

# Print summary of metrics including each fold's metrics
print("\nSummary of Metrics:")
for i, metrics in enumerate(fold_metrics_summary):
    print(f"\nFold {i + 1} Metrics:")
    print(f"Accuracy: {metrics[0]:.4f}")
    print(f"Precision: {metrics[1]:.4f}")
    print(f"Recall: {metrics[2]:.4f}")
    print(f"F1 Score: {metrics[3]:.4f}")
    print(f"Specificity: {metrics[4]:.4f}")
    print(f"AUROC: {metrics[5]:.4f}")
    print(f"AUPRC: {metrics[6]:.4f}")
    print(f"MCC: {mcc_values[i]:.4f}")

    
metrics_keys = ['Accuracy', 'Precision', 'Recall', 'F1 Score', 'Specificity', 'AUROC', 'AUPRC']

# Initialize the dictionary to accumulate fold and test set metrics
avg_metrics = {
    'Accuracy': 0,
    'Precision': 0,
    'Recall': 0,
    'F1 Score': 0,
    'Specificity': 0,
    'AUROC': 0,
    'AUPRC': 0
}

# Accumulate metrics for folds
for fold_metrics in fold_metrics_summary:
    accuracy, precision, recall, f1, specificity, auroc, auprc = fold_metrics
    avg_metrics['Accuracy'] += accuracy #done
    avg_metrics['Precision'] += precision #done
    avg_metrics['Recall'] += recall #done
    avg_metrics['F1 Score'] += f1 #done
    avg_metrics['Specificity'] += specificity #done
    avg_metrics['AUROC'] += auroc #done
    avg_metrics['AUPRC'] += auprc #done

    
# Add the test set metrics to the accumulated averages
avg_metrics['Accuracy'] += test_accuracy
avg_metrics['Precision'] += test_precision
avg_metrics['Recall'] += test_recall
avg_metrics['F1 Score'] += test_f1
avg_metrics['Specificity'] += test_specificity
avg_metrics['AUROC'] += test_auroc
avg_metrics['AUPRC'] += test_auprc



for key in metrics_keys:
    avg_metrics[key] /= 5

print("\nAverage Metrics Across Folds and Test Set:")
print(f"Average Accuracy: {avg_metrics['Accuracy']:.4f}")
print(f"Average Precision: {avg_metrics['Precision']:.4f}")
print(f"Average Recall: {avg_metrics['Recall']:.4f}")
print(f"Average F1 Score: {avg_metrics['F1 Score']:.4f}")
print(f"Average Specificity: {avg_metrics['Specificity']:.4f}")
print(f"Average AUROC: {avg_metrics['AUROC']:.4f}")
print(f"Average AUPRC: {avg_metrics['AUPRC']:.4f}")  # Print AUPRC as well

# Print confusion matrices for each fold
for i, cm in enumerate(confusion_matrices):
    print(f"\nConfusion Matrix for Fold {i + 1}:\n{cm}")

# Final confusion matrix for the test set
final_confusion_matrix = confusion_matrix(all_y_true_test, all_y_pred_test)
print(f"\nFinal Confusion Matrix on Test Set:\n{final_confusion_matrix}")
```
